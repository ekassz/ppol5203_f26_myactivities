C. Use the `cols_keep` list to create a new dataset, `ca_df_clean` that removes those two columns. Print the head

D. Write a user-defined function, `subset_onerace` that takes in three arguments:
- A string for a racial/ethnic group
- A string representing the name of the race/ethnicity column
- A dataframe

And subset the data to rows of data for that race/ethnicity

```python
## your code here
def subset_onerace(race, race_name, df):
    return df[df[race_name] == race]
```

E. Use that user-defined function to return the rows for `American Indian or Alaska Native`, ordered by year from lowest to highest. Has the uninsurance rate among children gone up or down over time?
